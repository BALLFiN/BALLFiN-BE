```python
# 특정 컬럼(예: 'col_name')에 NaN이 있는 행 제거
df = df.dropna(subset=['summary'])
```

```python
len(df)
```

```python
# 매핑 딕셔너리 정의
mapping = {
    "부정": "negative",
    "중립": "neutral",
    "긍정": "positive"
}

# signal 컬럼 매핑 적용
df['signal'] = df['signal'].map(mapping)
```

```python
def parse_datetime(korean_time_str):
    try:
        return datetime.strptime(korean_time_str, "%Y-%m-%d %H:%M:%S")
    except ValueError:
        return datetime.strptime(korean_time_str, "%Y-%m-%d %H:%M")
```

# Start

```python
# # ✅ 변환 및 삽입
documents = []
for _, row in df.iterrows():
    doc = {
        "link": row["link_url"],
        "press": row["agency"],
        "published_at": parse_datetime(row["date"]),
        "title": row["title"],
        "image_url": row.get("image_url"),
        "related_companies": "000660",
        "views": 0,
        "summary": row["summary"],
        "impact": row['signal'],  # 'positive' 또는 'negative'
        "analysis": row['detail'],
        "content": row['content']
    }
    documents.append(doc)
    
print(documents[0])

news_collection.insert_many(documents)
print("✅ 뉴스 데이터 삽입 완료")
```
